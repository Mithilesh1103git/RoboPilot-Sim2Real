## Submit a task
This requires `NEBIUS_API_KEY` and `NEBIUS_MODEL` in `.env`, plus the Panda model assets.

```python
task_response = requests.post(
    f"{BASE_URL}/tasks",
    json={
        "instruction": "Pick up the red block and place it in the blue tray",
        "scenario": "panda_pick_place",
    },
    timeout=20,
)
task_response.status_code, task_response.json()
```
